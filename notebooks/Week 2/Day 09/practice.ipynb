```python
def count_up_to(n):
    i=1
    while i<=n:     
        yield i
        i+=1
for number in count_up_to(5):
    print(number)
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average()

        if avg >= 80:
            return "A Grade"
        elif avg >= 70:
            return "B Grade"
        elif avg >= 60:
            return "C Grade"
        else:
            return "Fail"


student = Student("Fatima", [85, 92, 78])

print("Name : " , student.name)
print("Marks : " , student.marks)
print("Average : " , student.average())
print("grade : " , student.grade())
```

```python
def shout(func):
    def wrapper(name):
        print("Hello", func(name).upper())
    return wrapper


@shout
def greet(name):
    return name
greet("Iqra Rehman Shaikh")
```

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(10))
```
